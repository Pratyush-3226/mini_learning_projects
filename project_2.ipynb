```python
import matplotlib.pyplot as plt
titanic=pd.read_csv('/content/Titanic.csv')
gender=titanic['Gender'].value_counts()
plt.bar((gender.index),(gender.values),color="silver")
plt.title("Male and Female Passengers")
plt.xlabel("Gender")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
titanic = pd.read_csv('/content/Titanic.csv')
a= titanic["PassengerId"]
b= titanic["Fare"]
plt.plot(a,b,color="purple")
plt.title("Fare Change Across Passengers")
plt.xlabel("Passenger ID")
plt.ylabel("Fare")
plt.show()
```

```python
a=titanic['Age']
plt.hist(a,color="green")
plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
p=titanic["Embarked"].value_counts()
plt.pie(p.values,labels=p.index)
plt.title("Passengers Boarded")
plt.xlabel("Passenger")
plt.grid()
plt.show()
```

```python
titanic = pd.read_csv('/content/Titanic.csv')
a=titanic['Age']
b=titanic['Fare']
plt.scatter(a,b,color="red")
plt.title("Relationship between Age and Fare")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

```python
titanic = pd.read_csv('/content/Titanic.csv')
pclass=titanic["Pclass"].value_counts()
plt.barh(pclass.index, pclass.values,color="brown")
plt.title("Passenger Class Distribution")
plt.xlabel("Number of Passengers")
plt.ylabel("Passenger Class")
plt.show()
```
